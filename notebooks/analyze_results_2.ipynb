## 2. Load ColabDesign results

````python
def load_colab_results(colab_dir: pathlib.Path, source: str) -> pd.DataFrame:
    model = 'af2_multimer' if 'multimer' in source else 'af2'
    rows = []
    for jf in sorted(colab_dir.glob('*.json')):
        proto = jf.stem
        data  = json.loads(jf.read_text())
        for i, entry in enumerate(data.get('results', [])):
            rows.append({
                'source'     : source,
                'model'      : model,
                'group'      : f'{source}/{proto}',
                'name'       : f'{source}/{proto}_{i}',
                'protocol'   : proto,
                'idx'        : i,
                'seq'        : entry.get('seq', ''),
                'loss_total' : entry.get('loss_total'),
                'loss_af'    : entry.get('loss_af'),
                'energy'     : entry.get('energy'),
                'colab_iptm' : entry.get('i_ptm'),
                'colab_ptm'  : entry.get('ptm'),
                'colab_plddt': entry.get('plddt'),
            })
    if not rows:
        print(f'  [empty] {source}: no results found in {colab_dir}')
        return pd.DataFrame()
    df = pd.DataFrame(rows)
    print(f'{source:30s}  model={model}  {len(df):4d} entries  ({df["protocol"].nunique()} protocols)')
    return df

frames = []
for run_name, run_dir in COLAB_DIRS.items():
    frames.append(load_colab_results(run_dir, run_name))

df = pd.concat([f for f in frames if not f.empty], ignore_index=True)
print(f'\nTotal: {len(df)} entries from {df["group"].nunique()} (source, protocol) groups')
df.head()
````

## 7. Sequence space — PCA + t-SNE

Each peptide is encoded as an integer vector of length 15 (one integer per position, 1–20 for the 20 standard amino acids).
PCA is applied first to decorrelate; t-SNE then maps to 2-D.

### 7.1 Parameters

````python
AA_ORDER   = 'ACDEFGHIKLMNPQRSTVWY'
AA_TO_INT  = {aa: i+1 for i, aa in enumerate(AA_ORDER)}

N_PCA        = 10
PERPLEXITY   = 30
N_ITER       = 1000
LEARNING_RATE = 'auto'
RANDOM_STATE  = 42
````

### 7.2 Encode sequences

````python
def encode_sequence(seq: str, aa_to_int: dict, length: int = 15) -> np.ndarray:
    vec = np.zeros(length, dtype=np.float32)
    for i, aa in enumerate(seq[:length]):
        vec[i] = aa_to_int.get(aa, 0)
    return vec

SEQ_LEN = int(df['seq'].str.len().mode()[0])
print(f'Sequence length detected: {SEQ_LEN}')

mask  = df['seq'].str.len() == SEQ_LEN
df_ok = df[mask].reset_index(drop=True)
print(f'Sequences kept: {len(df_ok)}/{len(df)}')

X_raw = np.stack([encode_sequence(s, AA_TO_INT, SEQ_LEN) for s in df_ok['seq']])
print(f'Feature matrix: {X_raw.shape}')
````

### 7.3 PCA — scree plot and projection

````python
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

pca_full = PCA(random_state=RANDOM_STATE)
pca_full.fit(X_scaled)

evr    = pca_full.explained_variance_ratio_
cumevr = np.cumsum(evr)
n_comps = len(evr)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.bar(range(1, n_comps+1), evr * 100, color='steelblue')
ax1.set_xlabel('PC'); ax1.set_ylabel('Explained variance (%)')
ax1.set_title('Scree plot')

ax2.plot(range(1, n_comps+1), cumevr * 100, marker='o', color='steelblue', ms=4)
if N_PCA is not None:
    ax2.axvline(N_PCA, color='tomato', linestyle='--',
                label=f'N_PCA={N_PCA}  ({cumevr[N_PCA-1]*100:.1f}% var)')
    ax2.legend()
ax2.set_xlabel('Number of PCs'); ax2.set_ylabel('Cumulative explained variance (%)')
ax2.set_title('Cumulative variance')
ax2.yaxis.set_major_formatter(mticker.PercentFormatter())

plt.suptitle('PCA on integer-encoded sequences', fontsize=13)
plt.tight_layout()
plt.show()

if N_PCA is not None and N_PCA < X_scaled.shape[1]:
    pca_reduce = PCA(n_components=N_PCA, random_state=RANDOM_STATE)
    X_pca = pca_reduce.fit_transform(X_scaled)
    print(f'PCA: {X_scaled.shape[1]} → {N_PCA} dims  '
          f'({cumevr[N_PCA-1]*100:.1f}% variance retained)')
else:
    X_pca = X_scaled
    print('PCA skipped — using full scaled matrix')
````

### 7.4 t-SNE

````python
tsne = TSNE(
    n_components  = 2,
    perplexity    = min(PERPLEXITY, len(X_pca) - 1),
    learning_rate = LEARNING_RATE,
    random_state  = RANDOM_STATE,
    init          = 'pca',
    verbose       = 1,
)
X_tsne = tsne.fit_transform(X_pca)
df_ok  = df_ok.assign(tsne_x=X_tsne[:, 0], tsne_y=X_tsne[:, 1])
print('t-SNE done. KL divergence:', round(tsne.kl_divergence_, 4))
````

## 8. Multi-criteria candidate selection — 48 sequences

Selects 48 candidates balancing four criteria:

| Criterion | Implementation |
|---|---|
| **AF3 score** | `af3_iptm_cross` (best of 5 AF3 models by `ranking_score`) |
| **AF2/ColabDesign score** | `colab_iptm` (internal interface iPTM during hallucination) |
| **Sequence diversity** | Greedy selection with minimum pairwise Hamming distance constraint |
| **Protocol diversity** | Tracked across all design runs; reported in the final table |

**Composite score** (on min-max-normalised metrics):
- Sequences *with* AF3 data: `score = 0.55 × norm(af3_iptm_cross) + 0.45 × norm(colab_iptm)`
- Sequences *without* AF3: `score = 0.90 × norm(colab_iptm)` (10 % discount — no independent structural validation)

**Greedy diversity selection**: candidates are picked in composite-score order; each new sequence must differ from all already-selected sequences by at least `MIN_HAMMING` positions.  The distance threshold is relaxed by 1 at a time if the quota cannot be filled.

````python
# ── Selection parameters ────────────────────────────────────────────────────
N_CANDIDATES   = 48
W_AF3          = 0.55   # weight on af3_iptm_cross (AF3-validated sequences only)
W_AF2          = 0.45   # weight on colab_iptm
AF3_ONLY_SCALE = 0.90   # discount for sequences with no AF3 prediction
MIN_HAMMING    = 3      # minimum pairwise Hamming distance for diversity

# ── Deduplicate: one row per unique sequence ─────────────────────────────────
# For each unique seq keep the row with the best colab_iptm;
# aggregate protocols / sources from all design runs that produced it.
def agg_seq_group(grp):
    best = grp.loc[grp['colab_iptm'].idxmax()]
    af3_cols = [c for c in grp.columns if c.startswith('af3_')]
    result = {
        'seq'         : best['seq'],
        'colab_iptm'  : best['colab_iptm'],
        'colab_ptm'   : best['colab_ptm'],
        'colab_plddt' : best['colab_plddt'],
        'energy'      : best['energy'],
        'loss_af'     : best['loss_af'],
        'protocols'   : sorted(grp['protocol'].unique().tolist()),
        'sources'     : sorted(grp['source'].unique().tolist()),
        'n_runs'      : len(grp),
    }
    for c in af3_cols:
        result[c] = best[c]
    return pd.Series(result)

df_uniq = (
    df_m.groupby('seq', group_keys=False)
    .apply(agg_seq_group)
    .reset_index(drop=True)
)
print(f'Unique sequences : {len(df_uniq)}')
print(f'  with AF3 data  : {df_uniq["af3_iptm"].notna().sum()}')

# ── Normalise scores ─────────────────────────────────────────────────────────
def minmax(s: pd.Series) -> pd.Series:
    lo, hi = s.min(), s.max()
    return (s - lo) / (hi - lo) if hi > lo else pd.Series(0.5, index=s.index)

df_uniq['norm_af2'] = minmax(df_uniq['colab_iptm'])

af3_mask = df_uniq['af3_iptm_cross'].notna()
df_uniq['norm_af3'] = np.nan
if af3_mask.sum() > 0:
    df_uniq.loc[af3_mask, 'norm_af3'] = minmax(df_uniq.loc[af3_mask, 'af3_iptm_cross']).values

# ── Composite score ──────────────────────────────────────────────────────────
df_uniq['composite'] = np.where(
    af3_mask,
    W_AF3 * df_uniq['norm_af3'].fillna(0) + W_AF2 * df_uniq['norm_af2'],
    AF3_ONLY_SCALE * df_uniq['norm_af2'],
)

print('\nComposite score statistics:')
print(df_uniq['composite'].describe().round(3))
print('\nTop 10 by composite score:')
display(
    df_uniq.nlargest(10, 'composite')
    [['seq', 'colab_iptm', 'af3_iptm_cross', 'composite', 'protocols']]
    .reset_index(drop=True)
)
````

````python
def _hamming(a: str, b: str) -> int:
    return sum(x != y for x, y in zip(a, b))


def greedy_diverse_select(df_cands: pd.DataFrame, n: int, min_dist: int) -> pd.DataFrame:
    """
    Greedy selection of n sequences maximising composite score subject to
    pairwise Hamming distance >= min_dist.  If the constraint cannot be
    satisfied the threshold is relaxed by 1 and the loop retries.
    """
    seqs  = df_cands['seq'].tolist()
    order = sorted(range(len(seqs)), key=lambda i: -df_cands['composite'].iloc[i])

    selected, sel_set = [], set()
    d_min = min_dist

    while len(selected) < n:
        prev = len(selected)
        for i in order:
            if i in sel_set:
                continue
            ok = (not selected) or (
                min(_hamming(seqs[i], seqs[j]) for j in selected) >= d_min
            )
            if ok:
                selected.append(i)
                sel_set.add(i)
            if len(selected) >= n:
                break
        if len(selected) >= n:
            break
        if len(selected) == prev:           # zero progress → relax
            if d_min == 0:
                break
            d_min -= 1
            print(f'  no progress → relaxing min_dist to {d_min}  ({len(selected)}/{n})')
        elif len(selected) < n:             # some progress but not enough → relax
            d_min -= 1
            print(f'  relaxing min_dist to {d_min}  ({len(selected)}/{n} selected)')

    print(f'Selected {len(selected)} sequences  (effective min_dist = {d_min})')
    return df_cands.iloc[selected].reset_index(drop=True)


# Exclude custom protocol before selection
_EXCLUDE_S8 = {'seqs_aff_alberto_darren_6best'}
df_uniq_filtered = df_uniq[
    ~df_uniq['protocols'].apply(lambda ps: bool(_EXCLUDE_S8 & set(ps)))
].reset_index(drop=True)
print(f'Pool for §8: {len(df_uniq_filtered)} sequences  '
      f'({len(df_uniq) - len(df_uniq_filtered)} excluded — custom protocol)')

df_cands_sorted = df_uniq_filtered.sort_values('composite', ascending=False).reset_index(drop=True)
df_candidates   = greedy_diverse_select(df_cands_sorted, N_CANDIDATES, MIN_HAMMING)

# Protocol and source representation
proto_list   = [p for ps in df_candidates['protocols'] for p in ps]
proto_counts = pd.Series(proto_list).value_counts()
src_list     = [s for ss in df_candidates['sources'] for s in ss]
src_counts   = pd.Series(src_list).value_counts()

print(f'\n{len(df_candidates)} candidates selected')
print(f'\nProtocol representation:\n{proto_counts.to_string()}')
print(f'\nSource representation:\n{src_counts.to_string()}')

all_protos = set(df_uniq_filtered.explode('protocols')['protocols'].unique())
missing    = all_protos - set(proto_counts.index)
if missing:
    print(f'\nProtocols with zero representation: {sorted(missing)}')
else:
    print('\nAll protocols represented in the selection.')
````

## 9. Transparent 3-tier selection — 48 sequences

Alternativa più manuale e trasparente alla selezione composita della sezione 8.

| Tier | N | Criterio |
|---|---|---|
| **T1** | 15 | Top AF3 `iptm_cross` — le migliori strutture validate indipendentemente |
| **T2** | 15 | Top AF2 `colab_iptm` con diversità di sequenza (Hamming ≥ `MIN_HAMMING`), escluse da T1 |
| **T3** | 18 | Top composite score, non presenti in T1/T2, con diversità rispetto a tutto il set selezionato |

Escluse le sequenze dal protocollo `seqs_aff_alberto_darren_6best` (custom/non design automatico).

````python
# ── Parameters ────────────────────────────────────────────────────────────────
N_T1 = 15
N_T2 = 15
N_T3 = 18
EXCLUDE_PROTO = {'seqs_aff_alberto_darren_6best'}
MIN_HAMMING_3T = 3

# ── Exclude custom protocol and build selection pool ──────────────────────────
df_pool = df_uniq[
    ~df_uniq['protocols'].apply(lambda ps: bool(EXCLUDE_PROTO & set(ps)))
].copy().reset_index(drop=True)

n_excl = len(df_uniq) - len(df_pool)
print(f'Pool: {len(df_pool)} sequences  ({n_excl} excluded — custom protocol)')
print(f'  with AF3 data: {df_pool["af3_iptm_cross"].notna().sum()}')

# ── Helper: greedy diversity-aware selection ──────────────────────────────────
def greedy_select_diverse(df_cands, n, score_col, fixed_seqs, min_dist):
    """
    Pick n rows from df_cands ordered by score_col (desc).
    Each selected sequence must be >= min_dist Hamming from:
      - all sequences in fixed_seqs (previous tiers)
      - all sequences already selected in this call
    Threshold is relaxed by 1 if it cannot be filled.
    """
    seqs   = df_cands['seq'].str.upper().tolist()
    fixed  = [s.upper() for s in fixed_seqs]
    order  = sorted(range(len(seqs)), key=lambda i: -df_cands[score_col].iloc[i])
    selected, sel_set = [], set()
    d_min  = min_dist

    while len(selected) < n:
        prev = len(selected)
        for i in order:
            if i in sel_set:
                continue
            against = fixed + [seqs[j] for j in selected]
            ok = (not against) or (min(_hamming(seqs[i], s) for s in against) >= d_min)
            if ok:
                selected.append(i)
                sel_set.add(i)
            if len(selected) >= n:
                break
        if len(selected) >= n:
            break
        if len(selected) == prev:
            if d_min == 0:
                break
            d_min -= 1
            print(f'  no progress → relaxing min_dist to {d_min}  ({len(selected)}/{n})')
        elif len(selected) < n:
            d_min -= 1
            print(f'  relaxing min_dist to {d_min}  ({len(selected)}/{n})')

    print(f'  → {len(selected)} sequences selected  (effective min_dist = {d_min})')
    return df_cands.iloc[selected].copy().reset_index(drop=True)


# ── Tier 1: top N_T1 by AF3 iptm_cross (no diversity constraint) ─────────────
tier1 = (
    df_pool[df_pool['af3_iptm_cross'].notna()]
    .sort_values('af3_iptm_cross', ascending=False)
    .head(N_T1)
    .copy()
    .reset_index(drop=True)
)
tier1['tier'] = 'T1_af3'

print(f'\n── Tier 1: top {N_T1} by AF3 iPTM cross ──')
display(
    tier1[['seq', 'af3_iptm_cross', 'af3_ranking_score', 'colab_iptm', 'composite', 'protocols']]
    .rename_axis('rank')
    .assign(rank=range(1, len(tier1)+1))
    .set_index('rank')
)
````

````python
# ── Tier 2: top N_T2 by colab_iptm + diversity, not in Tier 1 ────────────────
t1_seqs = tier1['seq'].str.upper().tolist()
t2_pool = (
    df_pool[~df_pool['seq'].str.upper().isin(set(t1_seqs))]
    .sort_values('colab_iptm', ascending=False)
    .reset_index(drop=True)
)
print(f'Tier 2 pool: {len(t2_pool)} sequences')

tier2 = greedy_select_diverse(t2_pool, N_T2, 'colab_iptm', t1_seqs, MIN_HAMMING_3T)
tier2['tier'] = 'T2_af2_diverse'

print(f'\n── Tier 2: top {N_T2} by AF2 iPTM + sequence diversity ──')
display(
    tier2[['seq', 'colab_iptm', 'af3_iptm_cross', 'composite', 'protocols']]
    .assign(rank=range(1, len(tier2)+1))
    .set_index('rank')
)
````

````python
# ── Tier 3: top N_T3 by composite, not in T1/T2, diverse from all selected ───
t12_seqs = t1_seqs + tier2['seq'].str.upper().tolist()
t3_pool  = (
    df_pool[~df_pool['seq'].str.upper().isin(set(t12_seqs))]
    .sort_values('composite', ascending=False)
    .reset_index(drop=True)
)
print(f'Tier 3 pool: {len(t3_pool)} sequences')

tier3 = greedy_select_diverse(t3_pool, N_T3, 'composite', t12_seqs, MIN_HAMMING_3T)
tier3['tier'] = 'T3_composite'

print(f'\n── Tier 3: top {N_T3} by composite score ──')
display(
    tier3[['seq', 'composite', 'colab_iptm', 'af3_iptm_cross', 'protocols']]
    .assign(rank=range(1, len(tier3)+1))
    .set_index('rank')
)
````

````python
# ── Visualisation: 3 tiers in AF2 vs AF3 space + t-SNE ───────────────────────
TIER_COLORS = {'T1_af3': 'steelblue', 'T2_af2_diverse': 'tomato', 'T3_composite': 'seagreen'}
TIER_LABELS = {
    'T1_af3'        : f'T1: top AF3 (n={len(tier1)})',
    'T2_af2_diverse': f'T2: top AF2 + diversity (n={len(tier2)})',
    'T3_composite'  : f'T3: composite (n={len(tier3)})',
}

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left: AF3 vs AF2
ax = axes[0]
pool_af3 = df_pool['af3_iptm_cross'].notna()
ax.scatter(df_pool.loc[pool_af3, 'colab_iptm'],
           df_pool.loc[pool_af3, 'af3_iptm_cross'],
           color='lightgrey', s=20, alpha=0.4, zorder=1, label='pool (w/ AF3)')
for tier_name, tier_df in [('T1_af3', tier1), ('T2_af2_diverse', tier2), ('T3_composite', tier3)]:
    sub_af3 = tier_df[tier_df['af3_iptm_cross'].notna()]
    sub_no  = tier_df[tier_df['af3_iptm_cross'].isna()]
    ax.scatter(sub_af3['colab_iptm'], sub_af3['af3_iptm_cross'],
               color=TIER_COLORS[tier_name], s=80, zorder=3, label=TIER_LABELS[tier_name])
    if len(sub_no):
        ax.scatter(sub_no['colab_iptm'], [0]*len(sub_no),
                   color=TIER_COLORS[tier_name], s=80, marker='^', zorder=3)
ax.set_xlabel('ColabDesign iPTM'); ax.set_ylabel('AF3 iPTM cross')
ax.set_title('AF3 vs AF2  (triangles = no AF3 data, y=0)')
ax.legend(fontsize=8)

# Right: t-SNE
ax = axes[1]
ax.scatter(df_ok['tsne_x'], df_ok['tsne_y'], color='lightgrey', s=20, alpha=0.4, zorder=1)
for tier_name, tier_df in [('T3_composite', tier3), ('T2_af2_diverse', tier2), ('T1_af3', tier1)]:
    t_seqs = set(tier_df['seq'].str.upper())
    pts = df_ok[df_ok['seq'].str.upper().isin(t_seqs)]
    ax.scatter(pts['tsne_x'], pts['tsne_y'],
               color=TIER_COLORS[tier_name], s=70, zorder=4, label=TIER_LABELS[tier_name])
ax.set_xlabel('t-SNE 1'); ax.set_ylabel('t-SNE 2')
ax.set_title('t-SNE — 3-tier selection'); ax.set_xticks([]); ax.set_yticks([])
ax.legend(fontsize=8)

plt.suptitle(f'3-tier transparent selection — {N_T1+N_T2+N_T3} candidates', fontsize=13)
plt.tight_layout()
plt.show()

# ── Final table + save ────────────────────────────────────────────────────────
df_tiered = pd.concat([tier1, tier2, tier3], ignore_index=True)

out_cols = ['tier', 'seq', 'af3_iptm_cross', 'af3_ranking_score',
            'colab_iptm', 'composite', 'energy', 'colab_plddt',
            'n_runs', 'protocols', 'sources']
out_cols = [c for c in out_cols if c in df_tiered.columns]

df_tiered_out = df_tiered[out_cols].copy()
df_tiered_out.index = range(1, len(df_tiered_out)+1)

pd.set_option('display.max_rows', 60)
print(f'=== {len(df_tiered_out)} candidates — 3-tier selection ===')
display(df_tiered_out)

# Save
df_save_t = df_tiered_out.copy()
for c in ['protocols', 'sources']:
    if c in df_save_t.columns:
        df_save_t[c] = df_save_t[c].apply(lambda x: '|'.join(x) if isinstance(x, list) else x)

OUT_TIERED = SCRIPT_DIR / 'results' / 'candidates_48_tiered.csv'
df_save_t.to_csv(OUT_TIERED, index=True, index_label='rank')
print(f'\nSaved → {OUT_TIERED}')
````
